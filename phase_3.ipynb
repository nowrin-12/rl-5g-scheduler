```python
import gymnasium as gym

env = gym.make("CartPole-v1")
obs, info = env.reset(seed=0)
print("State (what the agent sees):", obs)
print("Actions available:", env.action_space)

obs, reward, terminated, truncated, info = env.step(0)
print("New state:", obs)
print("Reward:", reward)
print("Episode over?", terminated or truncated)
```

```python
import numpy as np

def run_random(n_episodes=20):
    scores = []
    for ep in range(n_episodes):
        obs, info = env.reset(seed=ep)
        total, done = 0, False
        while not done:
            action = env.action_space.sample()
            obs, reward, terminated, truncated, info = env.step(action)
            total += reward
            done = terminated or truncated
        scores.append(total)
    return np.mean(scores)

print("Random agent average score:", run_random())
```

```python
from stable_baselines3 import PPO

model = PPO("MlpPolicy", env, verbose=0, seed=0)
model.learn(total_timesteps=30000)
```

```python
from stable_baselines3.common.evaluation import evaluate_policy

mean_score, std_score = evaluate_policy(model, env, n_eval_episodes=20)
print("Trained agent score:", mean_score, "+/-", std_score)
```

```python
from stable_baselines3 import DQN

dqn = DQN("MlpPolicy", env, verbose=0, seed=0)
dqn.learn(total_timesteps=50000)
print(evaluate_policy(dqn, env, n_eval_episodes=20))
```
